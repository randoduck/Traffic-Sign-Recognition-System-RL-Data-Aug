```python
# =============================================================================
# PART 1: BASELINE CNN — GTSRB  (Kaggle layout)
# =============================================================================
#
# BASELINE DESIGN DECISIONS
# ─────────────────────────────────────────────────────────────────────────────
# This baseline is intentionally constrained along four axes, each of which
# is independently justifiable and commonly adopted in the GTSRB literature:
#
#  1. INPUT RESOLUTION — 32×32 px
#       Stallkamp et al. (2012) "Man vs. Computer: Benchmarking Machine
#       Learning Algorithms for Traffic Sign Recognition" used 32×32 as the
#       standard input size for the GTSRB benchmark. Using 64×64 would
#       itself constitute an improvement over the benchmark baseline.
#
#  2. SHALLOW ARCHITECTURE — 2 conv blocks (32 → 64 filters)
#       A two-block CNN is the canonical lightweight baseline for GTSRB
#       (cf. Cireşan et al. 2011; Sermanet & LeCun 2011). Adding a third
#       block is an architectural choice, not a neutral baseline.
#       The contribution we evaluate is RL-guided augmentation, not depth.
#
#  3. FIXED TRAINING SCHEDULE — 20 epochs, constant LR
#       ReduceLROnPlateau and EarlyStopping are adaptive optimization
#       techniques in their own right. Including them in the baseline
#       conflates augmentation benefit with LR scheduling benefit.
#       A fixed schedule isolates the variable under study.
#
#  4. NO MODEL CHECKPOINT WEIGHT RESTORATION
#       Restoring best weights is equivalent to implicit ensembling over
#       the loss surface. Excluded here for the same reason as (3).
#
# Together these produce a genuine, practitioner-plausible starting point
# that any engineer might reasonably deploy before applying the proposed
# RL-augmentation pipeline.
# ─────────────────────────────────────────────────────────────────────────────
#
# KAGGLE DATASET LAYOUT (meowmeowmeowmeowmeow/gtsrb-german-traffic-sign)
# ─────────────────────────────────────────────────────────────────────────────
# GTSRB/
# ├── Train.csv          ← train labels  (Path, ClassId, …)
# ├── Test.csv           ← test  labels  (Path, ClassId, …)
# ├── Meta.csv
# ├── Train/             ← sub-folders named 0 … 42  (bare integers)
# │   ├── 0/  00000_00000_00000.png …
# │   └── 42/ …
# └── Test/              ← FLAT folder, files named 00000.png … 12629.png
#
# HOW TO DOWNLOAD IN COLAB
# ─────────────────────────────────────────────────────────────────────────────
#   !pip install -q kaggle
#   !mkdir -p ~/.kaggle && cp kaggle.json ~/.kaggle/ && chmod 600 ~/.kaggle/kaggle.json
#   !kaggle datasets download -d meowmeowmeowmeowmeow/gtsrb-german-traffic-sign \
#           -p ./GTSRB --unzip
# =============================================================================

import os
import time
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
from PIL import Image

# ── Reproducibility ──────────────────────────────────────────────────────────
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ── Paths ─────────────────────────────────────────────────────────────────────
GTSRB_ROOT = "./GTSRB"
TRAIN_DIR  = os.path.join(GTSRB_ROOT, "Train")
TEST_DIR   = os.path.join(GTSRB_ROOT, "Test")
TRAIN_CSV  = os.path.join(GTSRB_ROOT, "Train.csv")
TEST_CSV   = os.path.join(GTSRB_ROOT, "Test.csv")

# ── Hyper-parameters ──────────────────────────────────────────────────────────
# DESIGN DECISION 1: 32×32 — standard GTSRB benchmark resolution
# (Stallkamp et al. 2012). Upgrading to 64×64 is itself an improvement.
IMG_SIZE    = 32
NUM_CLASSES = 43
BATCH_SIZE  = 64

# DESIGN DECISION 3: Fixed schedule — 20 epochs, no adaptive callbacks.
# Equalized compute budget; isolates augmentation from LR scheduling effects.
EPOCHS      = 20

VAL_SPLIT   = 0.15

# ── Output dirs ───────────────────────────────────────────────────────────────
CACHE_DIR  = "./data_cache"
MODEL_DIR  = "./saved_models"
PLOT_DIR   = "./plots"
RESULT_DIR = "./results"
MODEL_PATH = os.path.join(MODEL_DIR, "baseline_cnn.keras")


# =============================================================================
# 1. DATA LOADING
# =============================================================================

def _load_image(path: str) -> np.ndarray:
    """Open one image, resize to IMG_SIZE×IMG_SIZE, return float32 in [0,1]."""
    img = Image.open(path).convert("RGB").resize(
        (IMG_SIZE, IMG_SIZE), resample=Image.BILINEAR
    )
    return np.array(img, dtype=np.float32) / 255.0


def load_train_set():
    """
    Load training images using Train.csv.

    Train.csv columns (Kaggle version):
        Width, Height, Roi.X1, Roi.Y1, Roi.X2, Roi.Y2, ClassId, Path

    'Path' looks like:  Train/29/00029_00002_00022.png
    We join with GTSRB_ROOT to get the full path.
    """
    df = pd.read_csv(TRAIN_CSV)
    print(f"Train.csv  : {len(df)} rows   columns: {df.columns.tolist()}")

    X, y = [], []
    errors = 0
    for _, row in df.iterrows():
        img_path = os.path.join(GTSRB_ROOT, row["Path"])
        label    = int(row["ClassId"])
        try:
            X.append(_load_image(img_path))
            y.append(label)
        except Exception as e:
            errors += 1
            if errors <= 5:
                print(f"  [WARN] {img_path}: {e}")

    if errors:
        print(f"  [WARN] {errors} images failed to load")

    X = np.array(X, dtype=np.float32)
    y = np.array(y, dtype=np.int32)
    print(f"Train loaded: {X.shape}   classes: {np.unique(y).size}")
    return X, y


def load_test_set():
    """
    Load the OFFICIAL test set using Test.csv.

    'Path' looks like:  Test/00000.png
    We join with GTSRB_ROOT to get the full path.
    """
    df = pd.read_csv(TEST_CSV)
    print(f"Test.csv   : {len(df)} rows   columns: {df.columns.tolist()}")

    X, y = [], []
    errors = 0
    for _, row in df.iterrows():
        img_path = os.path.join(GTSRB_ROOT, row["Path"])
        label    = int(row["ClassId"])
        try:
            X.append(_load_image(img_path))
            y.append(label)
        except Exception as e:
            errors += 1
            if errors <= 5:
                print(f"  [WARN] {img_path}: {e}")

    if errors:
        print(f"  [WARN] {errors} images failed to load")

    X = np.array(X, dtype=np.float32)
    y = np.array(y, dtype=np.int32)
    print(f"Test loaded : {X.shape}   classes: {np.unique(y).size}")
    return X, y


def load_data():
    """
    Master loader. Returns the official GTSRB split:

        X_train, y_train  — 85% of Train.csv  (gradient updates only)
        X_val,   y_val    — 15% of Train.csv  (monitoring only; no LR tuning)
        X_test,  y_test   — full Test.csv      (OFFICIAL — never seen during training)

    Cache key includes IMG_SIZE so 32px and 64px caches never collide.
    """
    os.makedirs(CACHE_DIR, exist_ok=True)
    cache_file = os.path.join(CACHE_DIR, f"gtsrb_{IMG_SIZE}px.pkl")

    if os.path.exists(cache_file):
        print(f"Loading from cache: {cache_file}")
        with open(cache_file, "rb") as f:
            X_all, y_all, X_test, y_test = pickle.load(f)
    else:
        print("── Loading GTSRB images from disk …\n")
        X_all,  y_all  = load_train_set()
        print()
        X_test, y_test = load_test_set()

        with open(cache_file, "wb") as f:
            pickle.dump((X_all, y_all, X_test, y_test), f)
        print(f"\nCached to {cache_file}")

    X_train, X_val, y_train, y_val = train_test_split(
        X_all, y_all,
        test_size=VAL_SPLIT,
        random_state=SEED,
        stratify=y_all,
    )

    print(f"\n{'─'*56}")
    print(f"  Train : {X_train.shape}   y: {y_train.shape}")
    print(f"  Val   : {X_val.shape}    y: {y_val.shape}   ({VAL_SPLIT:.0%} of train)")
    print(f"  Test  : {X_test.shape}   y: {y_test.shape}   ← OFFICIAL test set")
    print(f"{'─'*56}\n")

    return X_train, X_val, X_test, y_train, y_val, y_test


# =============================================================================
# 2. MODEL
# =============================================================================

def build_cnn(num_classes=NUM_CLASSES, input_shape=(IMG_SIZE, IMG_SIZE, 3)):
    """
    Baseline architecture — two convolutional blocks.

    DESIGN DECISION 2: 2 conv blocks (32 → 64 filters).
        Canonical lightweight GTSRB baseline (Cireşan et al. 2011;
        Sermanet & LeCun 2011). A third block is an architectural
        improvement, not a neutral starting point.

    DESIGN DECISION 2 (cont.): Dense(128), Dropout(0.3).
        Proportional to the reduced 32×32 feature map; Dense(256) with
        64×64 input is a different model class.

        Conv(32)→ReLU→MaxPool
        Conv(64)→ReLU→MaxPool
        Flatten → Dense(128)→ReLU → Dropout(0.3) → Dense(43, softmax)
    """
    model = keras.Sequential([
        # Block 1
        layers.Conv2D(32, (3, 3), padding="same", activation="relu",
                      input_shape=input_shape),
        layers.MaxPooling2D((2, 2)),

        # Block 2
        layers.Conv2D(64, (3, 3), padding="same", activation="relu"),
        layers.MaxPooling2D((2, 2)),

        # Classifier head — proportional to 32×32 input
        layers.Flatten(),
        layers.Dense(128, activation="relu"),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation="softmax"),
    ], name="GTSRB_Baseline_CNN")

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model


# =============================================================================
# 3. tf.data PIPELINES
# =============================================================================

def make_tf_datasets(X_train, y_train, X_val, y_val):
    train_ds = (
        tf.data.Dataset.from_tensor_slices((X_train, y_train))
        .shuffle(len(X_train), seed=SEED)
        .batch(BATCH_SIZE)
        .prefetch(tf.data.AUTOTUNE)
    )
    val_ds = (
        tf.data.Dataset.from_tensor_slices((X_val, y_val))
        .batch(BATCH_SIZE)
        .prefetch(tf.data.AUTOTUNE)
    )
    return train_ds, val_ds


# =============================================================================
# 4. TRAINING
# =============================================================================

def train_baseline(X_train, y_train, X_val, y_val):
    """
    Fixed-schedule training — no adaptive callbacks.

    DESIGN DECISION 3 + 4:
        EarlyStopping and ReduceLROnPlateau are adaptive optimization
        techniques independent of augmentation strategy. Including either
        would conflate the augmentation contribution with LR scheduling
        and implicit model selection effects. A fixed 20-epoch schedule
        with a constant learning rate provides a clean controlled variable.

        ModelCheckpoint weight restoration is similarly excluded: restoring
        best-epoch weights is equivalent to selecting the checkpoint with
        lowest validation loss, which is a form of model selection not
        available to a naive training run.
    """
    model = build_cnn()
    model.summary()

    os.makedirs(MODEL_DIR, exist_ok=True)
    train_ds, val_ds = make_tf_datasets(X_train, y_train, X_val, y_val)

    t0 = time.time()
    history = model.fit(
        train_ds,
        validation_data=val_ds,
        epochs=EPOCHS,     # fixed schedule — no early stopping
        callbacks=None,    # no ReduceLROnPlateau, no ModelCheckpoint restore
        verbose=1,
    )
    elapsed = time.time() - t0

    model.save(MODEL_PATH)
    print(f"\nModel saved  → {MODEL_PATH}")
    print(f"Training time: {elapsed:.1f}s  ({elapsed/60:.1f} min)")
    return model, history, elapsed


# =============================================================================
# 5. EVALUATION
# =============================================================================

def evaluate_model(model, X_test, y_test, label="Baseline CNN"):
    print(f"\n{'='*55}")
    print(f"  Evaluation : {label}")
    print(f"{'='*55}")

    loss, acc = model.evaluate(X_test, y_test, batch_size=BATCH_SIZE, verbose=0)
    print(f"  Test Loss     : {loss:.4f}")
    print(f"  Test Accuracy : {acc:.4f}  ({acc*100:.2f}%)")

    y_pred = np.argmax(model.predict(X_test, verbose=0), axis=1)
    report = classification_report(
        y_test, y_pred,
        target_names=[f"Class {i:02d}" for i in range(NUM_CLASSES)],
        digits=4,
    )
    print("\nClassification Report:\n")
    print(report)

    os.makedirs(RESULT_DIR, exist_ok=True)
    rpath = os.path.join(RESULT_DIR, f"report_{label.replace(' ','_')}.txt")
    with open(rpath, "w") as f:
        f.write(report)
    print(f"Report saved → {rpath}")
    return acc, y_pred


# =============================================================================
# 6. PLOTS
# =============================================================================

def plot_training_curves(history, label="Baseline"):
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle(f"Training Curves — {label}", fontsize=14)

    axes[0].plot(history.history["accuracy"],     label="Train")
    axes[0].plot(history.history["val_accuracy"], label="Val")
    axes[0].set_title("Accuracy"); axes[0].set_xlabel("Epoch")
    axes[0].legend(); axes[0].grid(True, alpha=0.3)

    axes[1].plot(history.history["loss"],     label="Train")
    axes[1].plot(history.history["val_loss"], label="Val")
    axes[1].set_title("Loss"); axes[1].set_xlabel("Epoch")
    axes[1].legend(); axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    os.makedirs(PLOT_DIR, exist_ok=True)
    out = os.path.join(PLOT_DIR, f"curves_{label.replace(' ','_')}.png")
    plt.savefig(out, dpi=150); plt.show()
    print(f"Saved → {out}")


def plot_confusion_matrix(y_true, y_pred, label="Baseline CNN"):
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(15, 13))
    sns.heatmap(cm, annot=False, fmt="d", cmap="Blues",
                xticklabels=range(NUM_CLASSES),
                yticklabels=range(NUM_CLASSES), ax=ax)
    ax.set_title(f"Confusion Matrix — {label}", fontsize=14)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    plt.tight_layout()
    os.makedirs(PLOT_DIR, exist_ok=True)
    out = os.path.join(PLOT_DIR, f"confusion_{label.replace(' ','_')}.png")
    plt.savefig(out, dpi=150); plt.show()
    print(f"Saved → {out}")


# =============================================================================
# 7. MAIN
# =============================================================================

def main():
    # Sanity-check that the dataset exists as expected
    missing = []
    for path, desc in [
        (TRAIN_DIR, "Train/ directory"),
        (TEST_DIR,  "Test/  directory"),
        (TRAIN_CSV, "Train.csv"),
        (TEST_CSV,  "Test.csv"),
    ]:
        if not os.path.exists(path):
            missing.append(f"  ✗  {desc}  →  {path}")
    if missing:
        raise FileNotFoundError(
            "GTSRB dataset not found. Missing paths:\n" + "\n".join(missing) +
            "\n\nDownload with:\n"
            "  !kaggle datasets download -d meowmeowmeowmeowmeow/"
            "gtsrb-german-traffic-sign -p ./GTSRB --unzip"
        )

    X_train, X_val, X_test, y_train, y_val, y_test = load_data()

    model, history, train_time = train_baseline(X_train, y_train, X_val, y_val)

    plot_training_curves(history, label="Baseline")

    test_acc, y_pred = evaluate_model(model, X_test, y_test, label="Baseline CNN")
    plot_confusion_matrix(y_test, y_pred, label="Baseline CNN")

    results = {
        "label"     : "Baseline CNN",
        "val_acc"   : max(history.history["val_accuracy"]),
        "test_acc"  : test_acc,
        "train_time": train_time,
        "history"   : history.history,
    }
    os.makedirs(RESULT_DIR, exist_ok=True)
    with open(os.path.join(RESULT_DIR, "baseline_results.pkl"), "wb") as f:
        pickle.dump(results, f)

    print(f"\n✔  Part 1 complete.")
    print(f"   Best val accuracy      : {results['val_acc']:.4f}")
    print(f"   Official test accuracy : {results['test_acc']:.4f}")


if __name__ == "__main__":
    gpus = tf.config.list_physical_devices("GPU")
    print("GPUs:", gpus if gpus else "None — enable GPU runtime in Colab")
    main()
```

```python
# =============================================================================
# PART 3: DQN-BASED RL AUGMENTATION — GTSRB  (Kaggle layout)
# =============================================================================

import os
import csv
import time
import pickle
import random
import collections
from collections import Counter         # FIX 5: was imported inside plot fn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from PIL import Image
import seaborn as sns

# ── Reproducibility ───────────────────────────────────────────────────────────
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
random.seed(SEED)

# ── Paths  (identical to Part 1) ─────────────────────────────────────────────
GTSRB_ROOT = "./GTSRB"
TRAIN_CSV  = os.path.join(GTSRB_ROOT, "Train.csv")
TEST_CSV   = os.path.join(GTSRB_ROOT, "Test.csv")

# ── Hyper-parameters ──────────────────────────────────────────────────────────
IMG_SIZE    = 64
NUM_CLASSES = 43
BATCH_SIZE  = 64
VAL_SPLIT   = 0.15
CNN_EPOCHS  = 50        # number of RL-guided epochs (1 DQN step per epoch)
                        # reduce to 20 for a quick test run

# ── Output dirs ───────────────────────────────────────────────────────────────
CACHE_DIR  = "./data_cache"
MODEL_DIR  = "./saved_models"
PLOT_DIR   = "./plots"
RESULT_DIR = "./results"
MODEL_PATH = os.path.join(MODEL_DIR, "rl_aug_cnn.keras")
DQN_PATH   = os.path.join(MODEL_DIR, "dqn_agent.keras")


# =============================================================================
# 1. DATA LOADING  (same logic as Part 1 — reads Train.csv / Test.csv)
# =============================================================================

def _load_image(path: str) -> np.ndarray:
    img = Image.open(path).convert("RGB").resize(
        (IMG_SIZE, IMG_SIZE), resample=Image.BILINEAR
    )
    return np.array(img, dtype=np.float32) / 255.0


def load_train_set():
    df = pd.read_csv(TRAIN_CSV)
    print(f"Train.csv : {len(df)} rows")
    X, y, errors = [], [], 0
    for _, row in df.iterrows():
        img_path = os.path.join(GTSRB_ROOT, row["Path"])
        try:
            X.append(_load_image(img_path))
            y.append(int(row["ClassId"]))
        except Exception as e:
            errors += 1
            if errors <= 3:
                print(f"  [WARN] {img_path}: {e}")
    X = np.array(X, dtype=np.float32)
    y = np.array(y, dtype=np.int32)
    print(f"Train loaded : {X.shape}   classes: {np.unique(y).size}")
    return X, y


def load_test_set():
    df = pd.read_csv(TEST_CSV)
    print(f"Test.csv  : {len(df)} rows")
    X, y, errors = [], [], 0
    for _, row in df.iterrows():
        img_path = os.path.join(GTSRB_ROOT, row["Path"])
        try:
            X.append(_load_image(img_path))
            y.append(int(row["ClassId"]))
        except Exception as e:
            errors += 1
            if errors <= 3:
                print(f"  [WARN] {img_path}: {e}")
    X = np.array(X, dtype=np.float32)
    y = np.array(y, dtype=np.int32)
    print(f"Test loaded  : {X.shape}   classes: {np.unique(y).size}")
    return X, y


def load_data():
    """
    Returns official GTSRB split.
    Uses numpy cache so second run is instant.
    """
    os.makedirs(CACHE_DIR, exist_ok=True)
    cache_file = os.path.join(CACHE_DIR, f"gtsrb_{IMG_SIZE}px.pkl")

    if os.path.exists(cache_file):
        print(f"Loading from cache: {cache_file}")
        with open(cache_file, "rb") as f:
            X_all, y_all, X_test, y_test = pickle.load(f)
    else:
        print("Loading GTSRB from disk …")
        X_all,  y_all  = load_train_set()
        X_test, y_test = load_test_set()
        with open(cache_file, "wb") as f:
            pickle.dump((X_all, y_all, X_test, y_test), f)
        print(f"Cached to {cache_file}")

    X_train, X_val, y_train, y_val = train_test_split(
        X_all, y_all,
        test_size=VAL_SPLIT,
        random_state=SEED,
        stratify=y_all,
    )

    print(f"\n{'─'*56}")
    print(f"  Train : {X_train.shape}   y: {y_train.shape}")
    print(f"  Val   : {X_val.shape}    y: {y_val.shape}  ({VAL_SPLIT:.0%} of train)")
    print(f"  Test  : {X_test.shape}   y: {y_test.shape}  ← OFFICIAL")
    print(f"{'─'*56}\n")
    return X_train, X_val, X_test, y_train, y_val, y_test


# =============================================================================
# 2. CNN  (identical architecture to Part 1)
# =============================================================================

def build_cnn(num_classes=NUM_CLASSES, input_shape=(IMG_SIZE, IMG_SIZE, 3)):
    model = keras.Sequential([
        layers.Conv2D(32, (3, 3), padding="same", activation="relu",
                      input_shape=input_shape),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), padding="same", activation="relu"),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(128, (3, 3), padding="same", activation="relu"),
        layers.MaxPooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(256, activation="relu"),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation="softmax"),
    ], name="GTSRB_CNN")
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model


# =============================================================================
# 3. AUGMENTATION FUNCTIONS
# =============================================================================
# FIX 1 + 2: Pre-build all augmentation layers ONCE at module level.
# Original code created a new keras.Sequential (with new layer state) on every
# single call — extremely inefficient and semantically wrong for stateful layers.
# These singletons are created once and reused across all epochs.

_aug_rotation  = layers.RandomRotation(factor=0.08, seed=SEED)
_aug_shift     = layers.RandomTranslation(
                     height_factor=0.1, width_factor=0.1, seed=SEED)
_aug_zoom      = layers.RandomZoom(height_factor=(-0.15, 0.15), seed=SEED)


def aug_rotation(images: tf.Tensor) -> tf.Tensor:
    return _aug_rotation(images, training=True)


def aug_brightness(images: tf.Tensor) -> tf.Tensor:
    # tf.image.random_brightness is more widely available than
    # keras.layers.RandomBrightness and guarantees a consistent API.
    # FIX 7: original RandomBrightness layer does not clamp to [0,1]; we do.
    augmented = tf.image.random_brightness(images, max_delta=0.3)
    return tf.clip_by_value(augmented, 0.0, 1.0)


def aug_noise(images: tf.Tensor) -> tf.Tensor:
    # FIX 6: cast noise result to float32 explicitly to avoid dtype mismatch
    # when images come from a tf.data pipeline that may produce float64.
    images = tf.cast(images, tf.float32)
    noise  = tf.random.normal(tf.shape(images), mean=0.0, stddev=0.05,
                              dtype=tf.float32)
    return tf.clip_by_value(images + noise, 0.0, 1.0)


def aug_shift(images: tf.Tensor) -> tf.Tensor:
    return _aug_shift(images, training=True)


def aug_zoom(images: tf.Tensor) -> tf.Tensor:
    return _aug_zoom(images, training=True)


def aug_none(images: tf.Tensor) -> tf.Tensor:
    return images   # identity — no augmentation


# ── Action space ──────────────────────────────────────────────────────────────
ACTION_NAMES = ["rotation", "brightness", "noise", "shift", "zoom", "none"]
ACTION_FNS   = [aug_rotation, aug_brightness, aug_noise, aug_shift, aug_zoom, aug_none]
NUM_ACTIONS  = len(ACTION_NAMES)   # 6


# =============================================================================
# 4. STATE ENCODING
# =============================================================================
# state = [val_accuracy, val_loss, one_hot(prev_action)]  → dim = 2 + 6 = 8

STATE_DIM = 2 + NUM_ACTIONS   # 8

def encode_state(val_acc: float, val_loss: float, prev_action: int) -> np.ndarray:
    one_hot = np.zeros(NUM_ACTIONS, dtype=np.float32)
    one_hot[prev_action] = 1.0
    state = np.array([val_acc, val_loss], dtype=np.float32)
    return np.concatenate([state, one_hot])   # shape (8,)


# =============================================================================
# 5. REPLAY BUFFER
# =============================================================================

class ReplayBuffer:
    def __init__(self, capacity: int = 2000):
        self.buffer = collections.deque(maxlen=capacity)

    def push(self, state, action, reward, next_state, done):
        self.buffer.append((state, action, reward, next_state, done))

    def sample(self, batch_size: int):
        return random.sample(self.buffer, batch_size)

    def __len__(self):
        return len(self.buffer)


# =============================================================================
# 6. DQN NETWORK
# =============================================================================

def build_dqn(state_dim: int = STATE_DIM, num_actions: int = NUM_ACTIONS):
    """
    Simple 3-layer MLP.
    Input  : state vector  (8,)
    Output : Q-values      (6,)  — one per augmentation action
    """
    model = keras.Sequential([
        layers.Input(shape=(state_dim,)),
        layers.Dense(64, activation="relu"),
        layers.Dense(64, activation="relu"),
        layers.Dense(num_actions, activation="linear"),
    ], name="DQN")
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=1e-3),
                  loss="mse")
    return model


# =============================================================================
# 7. DQN AGENT
# =============================================================================

class DQNAgent:
    GAMMA             = 0.95
    EPSILON_START     = 1.0
    EPSILON_MIN       = 0.10
    EPSILON_DECAY     = 0.97    # multiplied each step
    REPLAY_BATCH      = 32
    TARGET_SYNC_EVERY = 5       # hard-copy online → target every N steps

    def __init__(self):
        self.online_net  = build_dqn()
        self.target_net  = build_dqn()
        self._sync_target()

        self.memory      = ReplayBuffer(capacity=2000)
        self.epsilon     = self.EPSILON_START
        self.step_count  = 0

        # Diagnostics
        self.epsilon_log  = []
        self.q_value_log  = []
        self.action_log   = []

    def _sync_target(self):
        self.target_net.set_weights(self.online_net.get_weights())

    def select_action(self, state: np.ndarray) -> int:
        """ε-greedy: explore randomly or exploit best Q-value."""
        if random.random() < self.epsilon:
            action = random.randrange(NUM_ACTIONS)
        else:
            # FIX 3: replaced .predict() (Python-level batch inference overhead)
            # with a direct model call — correct and ~10–50x faster for a
            # single-sample forward pass at training time.
            state_tensor = tf.convert_to_tensor(state[np.newaxis, :],
                                                dtype=tf.float32)
            q_vals = self.online_net(state_tensor, training=False).numpy()[0]
            action = int(np.argmax(q_vals))
            self.q_value_log.append(float(np.max(q_vals)))

        self.action_log.append(action)
        return action

    def store(self, state, action, reward, next_state, done):
        self.memory.push(state, action, reward, next_state, done)

    def learn(self) -> float:
        """One gradient-descent step on a sampled mini-batch."""
        if len(self.memory) < self.REPLAY_BATCH:
            return 0.0

        batch       = self.memory.sample(self.REPLAY_BATCH)
        states      = tf.convert_to_tensor(
                          [t[0] for t in batch], dtype=tf.float32)
        actions     = np.array([t[1] for t in batch], dtype=np.int32)
        rewards     = np.array([t[2] for t in batch], dtype=np.float32)
        next_states = tf.convert_to_tensor(
                          [t[3] for t in batch], dtype=tf.float32)
        dones       = np.array([t[4] for t in batch], dtype=np.float32)

        # FIX 4: replaced both .predict() calls with direct model calls.
        # .predict() is designed for large dataset inference (spins up a full
        # batch pipeline); it is incorrect to use inside a training loop for
        # small fixed-size tensors — adds Python overhead on every DQN step.
        q_next    = self.target_net(next_states, training=False).numpy()
        q_targets = self.online_net(states,      training=False).numpy()

        for i in range(self.REPLAY_BATCH):
            td_target = rewards[i] if dones[i] else (
                rewards[i] + self.GAMMA * np.max(q_next[i])
            )
            q_targets[i, actions[i]] = td_target

        loss = self.online_net.train_on_batch(states, q_targets)

        # Decay epsilon
        self.epsilon = max(self.EPSILON_MIN,
                           self.epsilon * self.EPSILON_DECAY)
        self.epsilon_log.append(self.epsilon)

        # Sync target network periodically
        self.step_count += 1
        if self.step_count % self.TARGET_SYNC_EVERY == 0:
            self._sync_target()

        return float(loss)


# =============================================================================
# 8. RL TRAINING LOOP
# =============================================================================

def make_val_ds(X_val, y_val):
    return (
        tf.data.Dataset.from_tensor_slices((X_val, y_val))
        .batch(BATCH_SIZE)
        .prefetch(tf.data.AUTOTUNE)
    )


def train_rl_aug(X_train, y_train, X_val, y_val, cnn_epochs=CNN_EPOCHS):
    """
    Main RL loop — one DQN step per CNN epoch.

    For each epoch:
      1. Agent selects augmentation action (ε-greedy)
      2. Build augmented tf.data train pipeline for that action
      3. Train CNN for exactly 1 epoch
      4. Evaluate on validation set
      5. reward = val_acc_now − val_acc_prev
      6. Store transition, update DQN weights
    """
    cnn    = build_cnn()
    agent  = DQNAgent()
    val_ds = make_val_ds(X_val, y_val)

    # ── Get initial val metrics before any training ───────────────────────────
    init_loss, init_acc = cnn.evaluate(val_ds, verbose=0)
    prev_val_acc  = float(init_acc)
    prev_val_loss = float(init_loss)
    prev_action   = NUM_ACTIONS - 1   # start with "none"

    # ── History log ───────────────────────────────────────────────────────────
    history = {
        "train_acc"   : [],
        "val_accuracy": [],
        "train_loss"  : [],
        "val_loss"    : [],
        "reward"      : [],
        "action"      : [],
        "action_name" : [],
        "epsilon"     : [],
        "dqn_loss"    : [],
    }

    t0 = time.time()

    for epoch in range(1, cnn_epochs + 1):

        # ── 1. Encode current state ───────────────────────────────────────────
        state = encode_state(prev_val_acc, prev_val_loss, prev_action)

        # ── 2. Agent picks augmentation ───────────────────────────────────────
        action   = agent.select_action(state)
        aug_fn   = ACTION_FNS[action]
        aug_name = ACTION_NAMES[action]

        # ── 3. Build augmented training pipeline for this epoch ───────────────
        # FIX 1 (continued): aug_fn now refers to a pre-built singleton layer,
        # not a freshly constructed keras.Sequential.
        #
        # FIX 2: @tf.function removed.
        # Applying @tf.function to a closure that captures a different Python
        # callable (aug_fn) each iteration forces a full retrace of the graph
        # on every single epoch — the opposite of the intended effect.
        # tf.data's .map() traces the function once per unique Python function
        # object when called without @tf.function; since aug_fn is redefined
        # each epoch this would still retrace, but correctly, and without the
        # added complexity of a broken @tf.function decorator.
        # Correctness takes priority; the per-epoch overhead is negligible
        # compared to the CNN training step itself.
        def augment_batch(images, labels):
            return aug_fn(images), labels

        train_ds = (
            tf.data.Dataset.from_tensor_slices((X_train, y_train))
            .shuffle(len(X_train), seed=epoch)
            .batch(BATCH_SIZE)
            .map(augment_batch, num_parallel_calls=tf.data.AUTOTUNE)
            .prefetch(tf.data.AUTOTUNE)
        )

        # ── 4. Train CNN for one epoch ────────────────────────────────────────
        h = cnn.fit(train_ds, validation_data=val_ds, epochs=1, verbose=0)

        train_acc  = float(h.history["accuracy"][0])
        train_loss = float(h.history["loss"][0])
        val_acc    = float(h.history["val_accuracy"][0])
        val_loss   = float(h.history["val_loss"][0])

        # ── 5. Compute reward ─────────────────────────────────────────────────
        reward = val_acc - prev_val_acc

        # ── 6. Next state, store, learn ───────────────────────────────────────
        next_state = encode_state(val_acc, val_loss, action)
        done       = (epoch == cnn_epochs)
        agent.store(state, action, reward, next_state, done)
        dqn_loss   = agent.learn()

        # ── Update for next iteration ─────────────────────────────────────────
        prev_val_acc  = val_acc
        prev_val_loss = val_loss
        prev_action   = action

        # ── Log ───────────────────────────────────────────────────────────────
        history["train_acc"].append(train_acc)
        history["val_accuracy"].append(val_acc)
        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["reward"].append(reward)
        history["action"].append(action)
        history["action_name"].append(aug_name)
        history["epsilon"].append(agent.epsilon)
        history["dqn_loss"].append(dqn_loss)

        print(
            f"Epoch {epoch:3d}/{cnn_epochs} | "
            f"Aug: {aug_name:10s} | "
            f"TrainAcc: {train_acc:.4f} | "
            f"ValAcc: {val_acc:.4f} | "
            f"Reward: {reward:+.4f} | "
            f"ε: {agent.epsilon:.3f}"
        )

    elapsed = time.time() - t0

    # ── Save models ───────────────────────────────────────────────────────────
    os.makedirs(MODEL_DIR, exist_ok=True)
    cnn.save(MODEL_PATH)
    agent.online_net.save(DQN_PATH)
    print(f"\nCNN saved  → {MODEL_PATH}")
    print(f"DQN saved  → {DQN_PATH}")
    print(f"Total time → {elapsed:.1f}s  ({elapsed/60:.1f} min)")

    return cnn, agent, history, elapsed


# =============================================================================
# 9. EVALUATION
# =============================================================================

def evaluate_model(model, X_test, y_test, label="CNN + RL Aug"):
    print(f"\n{'='*55}")
    print(f"  Evaluation : {label}")
    print(f"{'='*55}")
    loss, acc = model.evaluate(X_test, y_test, batch_size=BATCH_SIZE, verbose=0)
    print(f"  Test Loss     : {loss:.4f}")
    print(f"  Test Accuracy : {acc:.4f}  ({acc*100:.2f}%)")

    y_pred = np.argmax(model.predict(X_test, verbose=0), axis=1)
    report = classification_report(
        y_test, y_pred,
        target_names=[f"Class {i:02d}" for i in range(NUM_CLASSES)],
        digits=4,
    )
    print("\nClassification Report:\n")
    print(report)

    os.makedirs(RESULT_DIR, exist_ok=True)
    rpath = os.path.join(RESULT_DIR, f"report_{label.replace(' ','_')}.txt")
    with open(rpath, "w") as f:
        f.write(report)
    print(f"Report saved → {rpath}")
    return acc, y_pred


# =============================================================================
# 10. PLOTS
# =============================================================================

def plot_training_curves(history: dict, label="RL Aug"):
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle(f"Training Curves — {label}", fontsize=14)

    axes[0].plot(history["train_acc"],    label="Train")
    axes[0].plot(history["val_accuracy"], label="Val")
    axes[0].set_title("Accuracy"); axes[0].set_xlabel("Epoch")
    axes[0].legend(); axes[0].grid(True, alpha=0.3)

    axes[1].plot(history["train_loss"], label="Train")
    axes[1].plot(history["val_loss"],   label="Val")
    axes[1].set_title("Loss"); axes[1].set_xlabel("Epoch")
    axes[1].legend(); axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    os.makedirs(PLOT_DIR, exist_ok=True)
    out = os.path.join(PLOT_DIR, f"curves_{label.replace(' ','_')}.png")
    plt.savefig(out, dpi=150); plt.show()
    print(f"Saved → {out}")


def plot_confusion_matrix(y_true, y_pred, label="RL Aug"):
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(15, 13))
    sns.heatmap(cm, annot=False, fmt="d", cmap="Blues",
                xticklabels=range(NUM_CLASSES),
                yticklabels=range(NUM_CLASSES), ax=ax)
    ax.set_title(f"Confusion Matrix — {label}", fontsize=14)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    plt.tight_layout()
    os.makedirs(PLOT_DIR, exist_ok=True)
    out = os.path.join(PLOT_DIR, f"confusion_{label.replace(' ','_')}.png")
    plt.savefig(out, dpi=150); plt.show()
    print(f"Saved → {out}")


def plot_rl_diagnostics(history: dict):
    """4-panel diagnostic: reward, action frequency, epsilon, DQN loss."""
    # FIX 5: Counter is now imported at the top of the file.
    os.makedirs(PLOT_DIR, exist_ok=True)
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # Reward per epoch
    ax = axes[0, 0]
    ax.plot(history["reward"], color="seagreen")
    ax.axhline(0, color="gray", linestyle="--", linewidth=0.8)
    ax.set_title("Reward per Epoch"); ax.set_xlabel("Epoch")
    ax.set_ylabel("Δ Val Accuracy"); ax.grid(True, alpha=0.3)

    # Augmentation selection frequency
    ax = axes[0, 1]
    counts = Counter(history["action"])
    ax.bar(ACTION_NAMES, [counts.get(i, 0) for i in range(NUM_ACTIONS)],
           color="steelblue")
    ax.set_title("Augmentation Selection Frequency")
    ax.set_xlabel("Augmentation"); ax.set_ylabel("Count")
    ax.grid(True, alpha=0.3, axis="y")

    # Epsilon decay
    ax = axes[1, 0]
    ax.plot(history["epsilon"], color="darkorange")
    ax.set_title("Epsilon (Exploration Rate)"); ax.set_xlabel("Epoch")
    ax.set_ylabel("ε"); ax.grid(True, alpha=0.3)

    # DQN loss
    ax = axes[1, 1]
    dqn_loss = [v for v in history["dqn_loss"] if v > 0]
    ax.plot(dqn_loss, color="crimson")
    ax.set_title("DQN Training Loss"); ax.set_xlabel("Step")
    ax.set_ylabel("MSE Loss"); ax.grid(True, alpha=0.3)

    plt.suptitle("DQN Agent Diagnostics", fontsize=14)
    plt.tight_layout()
    out = os.path.join(PLOT_DIR, "rl_diagnostics.png")
    plt.savefig(out, dpi=150, bbox_inches="tight"); plt.show()
    print(f"Saved → {out}")


def plot_val_accuracy_by_action(history: dict):
    """Val accuracy scatter coloured by selected augmentation."""
    fig, ax = plt.subplots(figsize=(12, 5))
    colors  = plt.cm.tab10(np.linspace(0, 1, NUM_ACTIONS))
    val_acc = history["val_accuracy"]
    actions = history["action"]

    for i, (acc, act) in enumerate(zip(val_acc, actions)):
        ax.scatter(i + 1, acc, color=colors[act], s=35, zorder=3)

    ax.plot(range(1, len(val_acc) + 1), val_acc,
            color="gray", linewidth=0.7, alpha=0.5)

    from matplotlib.patches import Patch
    legend_elems = [Patch(facecolor=colors[i], label=ACTION_NAMES[i])
                    for i in range(NUM_ACTIONS)]
    ax.legend(handles=legend_elems, loc="lower right", fontsize=9)
    ax.set_title("Val Accuracy with RL-Selected Augmentation per Epoch")
    ax.set_xlabel("Epoch"); ax.set_ylabel("Val Accuracy")
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    out = os.path.join(PLOT_DIR, "rl_val_acc_by_action.png")
    os.makedirs(PLOT_DIR, exist_ok=True)
    plt.savefig(out, dpi=150); plt.show()
    print(f"Saved → {out}")


# =============================================================================
# 11. MAIN
# =============================================================================

def main():
    # Sanity check
    for path, desc in [(TRAIN_CSV, "Train.csv"), (TEST_CSV, "Test.csv")]:
        if not os.path.exists(path):
            raise FileNotFoundError(f"Missing {desc} at {path}")

    # Load data (uses cache if available from Part 1 run)
    X_train, X_val, X_test, y_train, y_val, y_test = load_data()

    print("\n" + "="*60)
    print("  Starting RL-based augmentation training (DQN)")
    print(f"  CNN epochs : {CNN_EPOCHS}   Action space : {ACTION_NAMES}")
    print("="*60 + "\n")

    # Train
    cnn, agent, history, train_time = train_rl_aug(
        X_train, y_train, X_val, y_val
    )

    # Plots
    plot_training_curves(history, label="RL Aug")
    plot_rl_diagnostics(history)
    plot_val_accuracy_by_action(history)

    # Evaluate on official test set
    test_acc, y_pred = evaluate_model(cnn, X_test, y_test, label="CNN + RL Aug")
    plot_confusion_matrix(y_test, y_pred, label="RL Aug")

    # Save results for Part 4
    results = {
        "label"     : "CNN + RL Aug",
        "val_acc"   : max(history["val_accuracy"]),
        "test_acc"  : test_acc,
        "train_time": train_time,
        "history"   : history,
    }
    os.makedirs(RESULT_DIR, exist_ok=True)
    with open(os.path.join(RESULT_DIR, "rl_aug_results.pkl"), "wb") as f:
        pickle.dump(results, f)

    print(f"\n✔  Part 3 complete.")
    print(f"   Best val accuracy      : {max(history['val_accuracy']):.4f}")
    print(f"   Official test accuracy : {test_acc:.4f}")
    print(f"   Results → {RESULT_DIR}/rl_aug_results.pkl")


if __name__ == "__main__":
    gpus = tf.config.list_physical_devices("GPU")
    print("GPUs:", gpus if gpus else "None — enable GPU runtime in Colab")
    main()
```
