```python
# FIXED AUGMEN
# =============================================================================
# PART 2: FIXED AUGMENTATION CNN — GTSRB  (Kaggle layout)
# =============================================================================

import os
import time
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
from PIL import Image

# ── Reproducibility ───────────────────────────────────────────────────────────
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ── Paths ─────────────────────────────────────────────────────────────────────
GTSRB_ROOT = "./GTSRB"
TRAIN_CSV  = os.path.join(GTSRB_ROOT, "Train.csv")
TEST_CSV   = os.path.join(GTSRB_ROOT, "Test.csv")

# ── Hyper-parameters ──────────────────────────────────────────────────────────
IMG_SIZE    = 64
NUM_CLASSES = 43
BATCH_SIZE  = 64
EPOCHS      = 30
VAL_SPLIT   = 0.15

# ── Output dirs ───────────────────────────────────────────────────────────────
CACHE_DIR  = "./data_cache"
MODEL_DIR  = "./saved_models"
PLOT_DIR   = "./plots"
RESULT_DIR = "./results"
MODEL_PATH = os.path.join(MODEL_DIR, "fixed_aug_cnn.keras")


# =============================================================================
# 1. DATA LOADING  (identical to Part 1)
# =============================================================================

def _load_image(path: str) -> np.ndarray:
    img = Image.open(path).convert("RGB").resize(
        (IMG_SIZE, IMG_SIZE), resample=Image.BILINEAR
    )
    return np.array(img, dtype=np.float32) / 255.0


def load_train_set():
    df = pd.read_csv(TRAIN_CSV)
    print(f"Train.csv : {len(df)} rows")
    X, y, errors = [], [], 0
    for _, row in df.iterrows():
        img_path = os.path.join(GTSRB_ROOT, row["Path"])
        try:
            X.append(_load_image(img_path))
            y.append(int(row["ClassId"]))
        except Exception as e:
            errors += 1
            if errors <= 3:
                print(f"  [WARN] {img_path}: {e}")
    X = np.array(X, dtype=np.float32)
    y = np.array(y, dtype=np.int32)
    print(f"Train loaded : {X.shape}   classes: {np.unique(y).size}")
    return X, y


def load_test_set():
    df = pd.read_csv(TEST_CSV)
    print(f"Test.csv  : {len(df)} rows")
    X, y, errors = [], [], 0
    for _, row in df.iterrows():
        img_path = os.path.join(GTSRB_ROOT, row["Path"])
        try:
            X.append(_load_image(img_path))
            y.append(int(row["ClassId"]))
        except Exception as e:
            errors += 1
            if errors <= 3:
                print(f"  [WARN] {img_path}: {e}")
    X = np.array(X, dtype=np.float32)
    y = np.array(y, dtype=np.int32)
    print(f"Test loaded  : {X.shape}   classes: {np.unique(y).size}")
    return X, y


def load_data():
    os.makedirs(CACHE_DIR, exist_ok=True)
    cache_file = os.path.join(CACHE_DIR, f"gtsrb_{IMG_SIZE}px.pkl")

    if os.path.exists(cache_file):
        print(f"Loading from cache: {cache_file}")
        with open(cache_file, "rb") as f:
            X_all, y_all, X_test, y_test = pickle.load(f)
    else:
        print("Loading GTSRB from disk …")
        X_all,  y_all  = load_train_set()
        X_test, y_test = load_test_set()
        with open(cache_file, "wb") as f:
            pickle.dump((X_all, y_all, X_test, y_test), f)
        print(f"Cached to {cache_file}")

    X_train, X_val, y_train, y_val = train_test_split(
        X_all, y_all,
        test_size=VAL_SPLIT,
        random_state=SEED,
        stratify=y_all,
    )

    print(f"\n{'─'*56}")
    print(f"  Train : {X_train.shape}   y: {y_train.shape}")
    print(f"  Val   : {X_val.shape}    y: {y_val.shape}  ({VAL_SPLIT:.0%} of train)")
    print(f"  Test  : {X_test.shape}   y: {y_test.shape}  ← OFFICIAL")
    print(f"{'─'*56}\n")
    return X_train, X_val, X_test, y_train, y_val, y_test


# =============================================================================
# 2. CNN  (identical to Part 1)
# =============================================================================

def build_cnn(num_classes=NUM_CLASSES, input_shape=(IMG_SIZE, IMG_SIZE, 3)):
    model = keras.Sequential([
        layers.Conv2D(32, (3, 3), padding="same", activation="relu",
                      input_shape=input_shape),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), padding="same", activation="relu"),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(128, (3, 3), padding="same", activation="relu"),
        layers.MaxPooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(256, activation="relu"),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation="softmax"),
    ], name="GTSRB_CNN")
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model


# =============================================================================
# 3. AUGMENTATION PIPELINE (FIXED FOR tf.function)
# =============================================================================

# IMPORTANT:
# Create augmentation layers ONCE outside the mapped function.
# If created inside augment_batch_fn, TensorFlow throws:
# "tf.function only supports singleton tf.Variables"

rotation_layer = tf.keras.layers.RandomRotation(
    factor=0.06,
    fill_mode="nearest",
    seed=SEED
)

translation_layer = tf.keras.layers.RandomTranslation(
    height_factor=0.08,
    width_factor=0.08,
    fill_mode="nearest",
    seed=SEED
)

zoom_layer = tf.keras.layers.RandomZoom(
    height_factor=(-0.10, 0.10),
    fill_mode="nearest",
    seed=SEED
)


def augment_batch_fn(images, labels):
    """
    Safe augmentation function for tf.data.Dataset.map()

    Input:
        images : batch tensor (B,H,W,3) in float32 [0,1]
        labels : batch labels

    Augmentations:
        - rotation
        - translation
        - zoom
        - brightness
        - gaussian noise
    """

    # rotation
    images = rotation_layer(images, training=True)

    # translation
    images = translation_layer(images, training=True)

    # zoom
    images = zoom_layer(images, training=True)

    # brightness adjustment
    images = tf.image.random_brightness(
        images,
        max_delta=0.2
    )
    images = tf.clip_by_value(images, 0.0, 1.0)

    # gaussian noise
    noise = tf.random.normal(
        shape=tf.shape(images),
        mean=0.0,
        stddev=0.03,
        dtype=tf.float32
    )

    images = tf.clip_by_value(
        images + noise,
        0.0,
        1.0
    )

    return images, labels


# =============================================================================
# 4. AUGMENTATION PREVIEW
# =============================================================================

def visualise_augmentations(X_sample: np.ndarray, n: int = 8):
    """Show original vs augmented side-by-side to verify images look correct."""
    sample  = X_sample[:n]
    batch   = tf.constant(sample)
    # Run augmentation once
    aug_batch, _ = augment_batch_fn(batch, tf.zeros(n, dtype=tf.int32))
    augmented = aug_batch.numpy()

    fig, axes = plt.subplots(2, n, figsize=(2 * n, 5))
    for i in range(n):
        axes[0, i].imshow(np.clip(sample[i], 0, 1))
        axes[0, i].axis("off")
        axes[0, i].set_title("Original", fontsize=7)

        axes[1, i].imshow(np.clip(augmented[i], 0, 1))
        axes[1, i].axis("off")
        axes[1, i].set_title("Augmented", fontsize=7)

    plt.suptitle("Fixed Augmentation Preview (corrected)", fontsize=12)
    plt.tight_layout()
    os.makedirs(PLOT_DIR, exist_ok=True)
    out = os.path.join(PLOT_DIR, "augmentation_preview.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    plt.show()
    print(f"Saved → {out}")

    # Quick sanity check — print pixel statistics
    print(f"\nPixel value stats after augmentation:")
    print(f"  min  : {augmented.min():.4f}  (should be ≥ 0)")
    print(f"  max  : {augmented.max():.4f}  (should be ≤ 1)")
    print(f"  mean : {augmented.mean():.4f}  (original mean: {sample.mean():.4f})")


# =============================================================================
# 5. tf.data PIPELINES
# =============================================================================

def make_augmented_tf_datasets(X_train, y_train, X_val, y_val):
    # Training: augment on-the-fly
    train_ds = (
        tf.data.Dataset.from_tensor_slices((X_train, y_train))
        .shuffle(len(X_train), seed=SEED)
        .batch(BATCH_SIZE)
        .map(augment_batch_fn, num_parallel_calls=tf.data.AUTOTUNE)
        .prefetch(tf.data.AUTOTUNE)
    )
    # Validation: never augmented
    val_ds = (
        tf.data.Dataset.from_tensor_slices((X_val, y_val))
        .batch(BATCH_SIZE)
        .prefetch(tf.data.AUTOTUNE)
    )
    return train_ds, val_ds


# =============================================================================
# 6. TRAINING
# =============================================================================

def train_fixed_aug(X_train, y_train, X_val, y_val):
    model = build_cnn()
    model.summary()

    os.makedirs(MODEL_DIR, exist_ok=True)
    callbacks = [
        keras.callbacks.EarlyStopping(
            monitor="val_accuracy", patience=7,
            restore_best_weights=True, verbose=1,
        ),
        keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss", factor=0.5,
            patience=3, min_lr=1e-6, verbose=1,
        ),
        keras.callbacks.ModelCheckpoint(
            filepath=os.path.join(MODEL_DIR, "fixed_aug_best.keras"),
            monitor="val_accuracy", save_best_only=True, verbose=0,
        ),
    ]

    train_ds, val_ds = make_augmented_tf_datasets(X_train, y_train, X_val, y_val)

    t0 = time.time()
    history = model.fit(
        train_ds,
        validation_data=val_ds,
        epochs=EPOCHS,
        callbacks=callbacks,
        verbose=1,
    )
    elapsed = time.time() - t0

    model.save(MODEL_PATH)
    print(f"\nModel saved  → {MODEL_PATH}")
    print(f"Training time: {elapsed:.1f}s  ({elapsed/60:.1f} min)")
    return model, history, elapsed


# =============================================================================
# 7. EVALUATION
# =============================================================================

def evaluate_model(model, X_test, y_test, label="CNN + Fixed Aug"):
    print(f"\n{'='*55}")
    print(f"  Evaluation : {label}")
    print(f"{'='*55}")
    loss, acc = model.evaluate(X_test, y_test, batch_size=BATCH_SIZE, verbose=0)
    print(f"  Test Loss     : {loss:.4f}")
    print(f"  Test Accuracy : {acc:.4f}  ({acc*100:.2f}%)")

    y_pred = np.argmax(model.predict(X_test, verbose=0), axis=1)
    report = classification_report(
        y_test, y_pred,
        target_names=[f"Class {i:02d}" for i in range(NUM_CLASSES)],
        digits=4,
    )
    print("\nClassification Report:\n")
    print(report)

    os.makedirs(RESULT_DIR, exist_ok=True)
    rpath = os.path.join(RESULT_DIR, f"report_{label.replace(' ','_')}.txt")
    with open(rpath, "w") as f:
        f.write(report)
    print(f"Report saved → {rpath}")
    return acc, y_pred


# =============================================================================
# 8. PLOTS
# =============================================================================

def plot_training_curves(history, label="Fixed Aug"):
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle(f"Training Curves — {label}", fontsize=14)

    axes[0].plot(history.history["accuracy"],     label="Train")
    axes[0].plot(history.history["val_accuracy"], label="Val")
    axes[0].set_title("Accuracy"); axes[0].set_xlabel("Epoch")
    axes[0].legend(); axes[0].grid(True, alpha=0.3)

    axes[1].plot(history.history["loss"],     label="Train")
    axes[1].plot(history.history["val_loss"], label="Val")
    axes[1].set_title("Loss"); axes[1].set_xlabel("Epoch")
    axes[1].legend(); axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    os.makedirs(PLOT_DIR, exist_ok=True)
    out = os.path.join(PLOT_DIR, f"curves_{label.replace(' ','_')}.png")
    plt.savefig(out, dpi=150); plt.show()
    print(f"Saved → {out}")


def plot_confusion_matrix(y_true, y_pred, label="Fixed Aug"):
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(15, 13))
    sns.heatmap(cm, annot=False, fmt="d", cmap="Blues",
                xticklabels=range(NUM_CLASSES),
                yticklabels=range(NUM_CLASSES), ax=ax)
    ax.set_title(f"Confusion Matrix — {label}", fontsize=14)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    plt.tight_layout()
    os.makedirs(PLOT_DIR, exist_ok=True)
    out = os.path.join(PLOT_DIR, f"confusion_{label.replace(' ','_')}.png")
    plt.savefig(out, dpi=150); plt.show()
    print(f"Saved → {out}")


def compare_with_baseline(baseline_results: dict, fixed_history,
                           fixed_test_acc: float, fixed_time: float):
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.plot(baseline_results["history"]["val_accuracy"],
            label="Baseline Val Acc", linestyle="--", color="steelblue")
    ax.plot(fixed_history.history["val_accuracy"],
            label="Fixed Aug Val Acc", linestyle="-", color="darkorange")
    ax.set_title("Validation Accuracy: Baseline vs Fixed Augmentation")
    ax.set_xlabel("Epoch"); ax.set_ylabel("Accuracy")
    ax.legend(); ax.grid(True, alpha=0.3)
    plt.tight_layout()
    out = os.path.join(PLOT_DIR, "comparison_baseline_vs_fixed.png")
    plt.savefig(out, dpi=150); plt.show()
    print(f"Saved → {out}")

    b = baseline_results
    print(f"\n{'='*52}")
    print(f"{'Metric':<25} {'Baseline':>12} {'Fixed Aug':>12}")
    print(f"{'─'*52}")
    print(f"{'Best Val Accuracy':<25} {b['val_acc']:>12.4f} "
          f"{max(fixed_history.history['val_accuracy']):>12.4f}")
    print(f"{'Test Accuracy':<25} {b['test_acc']:>12.4f} {fixed_test_acc:>12.4f}")
    print(f"{'Training Time (s)':<25} {b['train_time']:>12.1f} {fixed_time:>12.1f}")
    print(f"{'='*52}")


# =============================================================================
# 9. MAIN
# =============================================================================

def main():
    for path, desc in [(TRAIN_CSV, "Train.csv"), (TEST_CSV, "Test.csv")]:
        if not os.path.exists(path):
            raise FileNotFoundError(f"Missing {desc} at {path}")

    X_train, X_val, X_test, y_train, y_val, y_test = load_data()

    # Preview augmentation — verify images look correct before training
    print("Showing augmentation preview — check images are NOT black …")
    visualise_augmentations(X_train, n=8)

    # Train
    model, history, train_time = train_fixed_aug(X_train, y_train, X_val, y_val)

    # Plots
    plot_training_curves(history, label="Fixed Aug")

    # Evaluate on official test set
    test_acc, y_pred = evaluate_model(model, X_test, y_test,
                                       label="CNN + Fixed Aug")
    plot_confusion_matrix(y_test, y_pred, label="Fixed Aug")

    # Compare with baseline if Part 1 results exist
    baseline_path = os.path.join(RESULT_DIR, "baseline_results.pkl")
    if os.path.exists(baseline_path):
        with open(baseline_path, "rb") as f:
            baseline_results = pickle.load(f)
        compare_with_baseline(baseline_results, history, test_acc, train_time)
    else:
        print("\nBaseline results not found — run Part 1 first for comparison.")

    # Save results for Part 4
    results = {
        "label"     : "CNN + Fixed Aug",
        "val_acc"   : max(history.history["val_accuracy"]),
        "test_acc"  : test_acc,
        "train_time": train_time,
        "history"   : history.history,
    }
    os.makedirs(RESULT_DIR, exist_ok=True)
    with open(os.path.join(RESULT_DIR, "fixed_aug_results.pkl"), "wb") as f:
        pickle.dump(results, f)

    print(f"\n✔  Part 2 complete.")
    print(f"   Best val accuracy      : {results['val_acc']:.4f}")
    print(f"   Official test accuracy : {results['test_acc']:.4f}")


if __name__ == "__main__":
    gpus = tf.config.list_physical_devices("GPU")
    print("GPUs:", gpus if gpus else "None — enable GPU runtime in Colab")
    main()
```
